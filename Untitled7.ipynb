```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df_fraude = pd.read_csv('/content/creditcard.csv')

# Mostrar o tamanho real da base (Linhas vs Colunas)
print(f"O dataset possui {df_fraude.shape[0]:,} linhas e {df_fraude.shape[1]} colunas.")
```

```python
# Contagem exata de transações legítimas (0) e fraudes (1)
contagem = df_fraude['Class'].value_counts()
percentual = df_fraude['Class'].value_counts(normalize=True) * 100

print("--- Volume de Transações ---")
print(f"Legítimas (Classe 0): {contagem[0]:,} transações ({percentual[0]:.2f}%)")
print(f"Fraudes    (Classe 1): {contagem[1]:,} transações ({percentual[1]:.2f}%)")
```

Durante a análise exploratória, identifiquei que a base possui apenas 492 transações fraudulentas diante de mais de 284 mil transações legítimas. Isso representa meros 0.17% do total. Esse cenário reflete perfeitamente a realidade de uma operação de meios de pagamento: o desafio não é apenas achar a fraude, mas encontrá-la sem gerar fricção ou falsos positivos nos 99.83% de clientes bons.

```python
# Análise Estatística dos Valores (Amount) por Classe
print("--- Resumo Estatístico dos Valores de Transações ---")

print("\n[Transações Legítimas]")
print(df_fraude[df_fraude['Class'] == 0]['Amount'].describe())

print("\n[Transações Fraudulentas]")
print(df_fraude[df_fraude['Class'] == 1]['Amount'].describe())
```

```python
# Preparação dos dados / padronização
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_fraude['Amount_Scaled'] = scaler.fit_transform(df_fraude['Amount'].values.reshape(-1, 1))

df_limpo = df_fraude.drop(['Time', 'Amount'], axis=1)

# 4. Separar as variáveis preditoras (X) da resposta/alvo (y)
# X terá todas as características da transação, menos a resposta final ('Class')
X = df_limpo.drop('Class', axis=1)
# y terá apenas a resposta: é fraude (1) ou não (0)?
y = df_limpo['Class']

print("--- Dados preparados com sucesso! ---")
print(f"Formato da matriz X (Características): {X.shape}")
print(f"Formato do vetor y (Alvo): {y.shape}")
```

```python

from sklearn.model_selection import train_test_split

# Dividir os dados (será salvo 20% para o teste final)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

print("--- Divisão Executada com Sucesso! ---")
print(f"X_train (Dados de estudo da IA): {X_train.shape}")
print(f"X_test (Prova final da IA): {X_test.shape}")

# Verificar se a proporção de fraudes continuou igual nas duas partes
print("\nProporção de fraudes no Treino:")
print(y_train.value_counts(normalize=True) * 100)
```

```python
# (Regressão Logística)
from sklearn.linear_model import LogisticRegression

modelo_fraude = LogisticRegression(max_iter=1000, random_state=42)

# O TREINAMENTO: Aqui a IA analisa o X_train (dados) e o y_train (respostas)
# para descobrir as regras e padrões que diferenciam um cliente bom de um fraudador
print("Treinando a Inteligência Artificial... Por favor, aguarde.")
modelo_fraude.fit(X_train, y_train)
print("--- Treinamento Concluído com Sucesso! ---")

# 4. A PROVA FINAL: Avaliar dados da base teste
y_pred = modelo_fraude.predict(X_test)
print("A IA já fez as previsões para a base de teste!")
```

```python
# Avaliação
from sklearn.metrics import classification_report, confusion_matrix

print("--- Relatório de Desempenho da IA ---")
print(classification_report(y_test, y_pred))

print("\n--- Matriz de Confusão (Números Absolutos) ---")
print(confusion_matrix(y_test, y_pred))
```
